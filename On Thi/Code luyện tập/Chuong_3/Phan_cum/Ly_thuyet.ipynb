# 1. Phân loại và hồi quy: đều là học có giám sát

Tức là dữ liệu có:

```python
X = dữ liệu đầu vào
y = đáp án đúng
```

Vì có `y`, nên thường sẽ chia:

```python
x_train, x_test, y_train, y_test = train_test_split(X, y, ...)
```

## Điểm giống nhau

Cả **phân loại** và **hồi quy** đều có quy trình:

```text
1. Có X và y
2. Chia train/test
3. Tạo mô hình
4. fit(x_train, y_train)
5. predict(x_test)
6. So sánh y_test với dự đoán
```

---

# 2. Khác nhau giữa phân loại và hồi quy

## Phân loại

Dự đoán ra **nhãn/lớp**:

```text
0, 1, 2
Benign, Malignant
Có bệnh, Không bệnh
```

Dùng mô hình:

```python
RandomForestClassifier()
KNeighborsClassifier()
SVC()
LogisticRegression()
```

Đánh giá bằng:

```python
accuracy_score()
classification_report()
confusion_matrix()
f1_score()
```

Ví dụ:

```python
du_doan = mo_hinh.predict(x_test)
print(classification_report(y_test, du_doan))
```

---

## Hồi quy

Dự đoán ra **số liên tục**:

```text
giá nhà = 1.5 tỷ
điểm số = 8.5
chỉ số bệnh = 151.3
doanh thu = 200 triệu
```

Dùng mô hình:

```python
RandomForestRegressor()
LinearRegression()
KNeighborsRegressor()
SVR()
```

Đánh giá bằng:

```python
mean_absolute_error()
mean_squared_error()
r2_score()
```

Ví dụ:

```python
du_doan = mo_hinh.predict(x_test)

mae = mean_absolute_error(y_test, du_doan)
mse = mean_squared_error(y_test, du_doan)
r2 = r2_score(y_test, du_doan)
```

# 3. Phân cụm: học không giám sát

Phân cụm thì **không có y để học**.

Chỉ có:

```python
X = dữ liệu đầu vào
```

Không có:

```python
y_train
y_test
```

Nên thông thường **không cần `train_test_split`**.

Quy trình phân cụm thường là:

```text
1. Có X
2. Chuẩn hóa X nếu cần
3. Tạo mô hình phân cụm
4. fit(X)
5. Lấy nhãn cụm bằng labels_
6. Đánh giá bằng silhouette_score hoặc inertia
7. Vẽ biểu đồ cụm
```

Ví dụ:

```python
mo_hinh = KMeans(n_clusters=3, random_state=42)

mo_hinh.fit(X)

cum = mo_hinh.labels_
```

Hoặc viết gọn:

```python
cum = mo_hinh.fit_predict(X)
```

# 4. Bảng chốt dễ nhớ

| Nội dung                     | Phân loại                      | Hồi quy                 | Phân cụm                        |
| ---------------------------- | ------------------------------ | ----------------------- | ------------------------------- |
| Có `X` không?                | Có                             | Có                      | Có                              |
| Có `y` không?                | Có                             | Có                      | Không                           |
| Có `train_test_split` không? | Thường có                      | Thường có               | Thường không                    |
| Huấn luyện                   | `fit(x_train, y_train)`        | `fit(x_train, y_train)` | `fit(X)`                        |
| Dự đoán/lấy kết quả          | `predict(x_test)`              | `predict(x_test)`       | `labels_` hoặc `fit_predict(X)` |
| Kết quả                      | Lớp/nhãn                       | Số liên tục             | Cụm/nhóm                        |
| Đánh giá                     | Accuracy, F1, Confusion Matrix | MAE, MSE, RMSE, R2      | Silhouette Score, Inertia       |

---

# 5. Câu nhớ nhanh khi đi thi

Bạn nhớ 3 câu này là ăn điểm:

```text
Phân loại và hồi quy là học có giám sát vì có nhãn y.
Phân loại dự đoán nhãn lớp, hồi quy dự đoán giá trị số liên tục.
Phân cụm là học không giám sát, không có y, mô hình tự chia dữ liệu thành các cụm dựa trên độ giống nhau.
```

Nói ngắn gọn nhất:

```text
Phân loại: Có y, dự đoán lớp.
Hồi quy: Có y, dự đoán số.
Phân cụm: Không có y, tự chia nhóm.
```


````python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, adjusted_rand_score
from sklearn.decomposition import PCA

# 1. Tải dữ liệu
du_lieu = load_iris()

X = du_lieu.data
y = du_lieu.target

print("Kích thước X:", X.shape)
print("Tên đặc trưng:", du_lieu.feature_names)
print("Tên nhãn thật:", du_lieu.target_names)

# 2. Chuẩn hóa dữ liệu
chuan_hoa = StandardScaler()
X_chuan_hoa = chuan_hoa.fit_transform(X)

# 3. Tạo mô hình KMeans
mo_hinh = KMeans(n_clusters=3, random_state=42, n_init=10)

# 4. Phân cụm
cum = mo_hinh.fit_predict(X_chuan_hoa)

print("Nhãn cụm:")
print(cum)

# 5. Đánh giá
silhouette = silhouette_score(X_chuan_hoa, cum)
ari = adjusted_rand_score(y, cum)

print("Silhouette Score:", silhouette)
print("Adjusted Rand Index:", ari)

# 6. Giảm chiều bằng PCA để vẽ
pca = PCA(n_components=2)
X_2d = pca.fit_transform(X_chuan_hoa)

# 7. Vẽ biểu đồ scatter
plt.figure(figsize=(8, 6))

plt.scatter(X_2d[:, 0], X_2d[:, 1], c=cum)

plt.xlabel("PCA1")
plt.ylabel("PCA2")
plt.title("Phân cụm Iris bằng KMeans")

plt.show()
````
