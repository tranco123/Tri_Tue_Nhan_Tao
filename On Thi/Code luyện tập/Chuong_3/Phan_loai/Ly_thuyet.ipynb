## 1. Bài toán phân loại là gì?
### Hiểu đơn giản: Phân loại là bài toán dự đoán một đối tượng thuộc nhóm nào / lớp nào / loại nào.

| Bài toán        | Đầu vào                        | Đầu ra cần dự đoán    |
| --------------- | ------------------------------ | --------------------- |
| Hoa Iris        | Chiều dài, chiều rộng cánh hoa | Loại hoa              |
| Ung thư vú      | Các chỉ số tế bào              | Ác tính hay lành tính |
| Chữ số viết tay | Ảnh 8x8 pixel                  | Số 0, 1, 2, ..., 9    |
| Rượu vang       | Thành phần hóa học             | Loại rượu             |
