# 1. Bản chất ảnh trong máy tính

Trước khi học OpenCV, bạn phải hiểu:

> Ảnh trong máy tính thực chất là **ma trận số**.

Ví dụ ảnh xám:

```text
Ảnh xám = ma trận 2 chiều
Mỗi ô là một pixel
Giá trị pixel từ 0 đến 255
0 = đen
255 = trắng
```

Ví dụ đơn giản:

```text
[[  0,  50, 100],
 [120, 180, 220],
 [255, 200, 150]]
```

Còn ảnh màu thì có 3 kênh:

```text
Ảnh màu = ma trận 3 chiều
Chiều cao x chiều rộng x 3 kênh màu
```

Trong OpenCV, ảnh màu mặc định là:

```text
BGR
Blue - Green - Red
```

Không phải RGB như bình thường.

# 3. Các khái niệm học sâu cần hiểu

## ANN là gì?

ANN là **Artificial Neural Network**, tức là mạng nơ-ron nhân tạo.

Bạn hiểu đơn giản:

```text
ANN gồm nhiều lớp:
Input layer → Hidden layer → Output layer
```

Ví dụ:

```text
Input: dữ liệu đầu vào
Hidden: xử lý, học đặc trưng
Output: kết quả dự đoán
```

Câu học thuộc:

```text
ANN là mạng nơ-ron nhân tạo gồm lớp đầu vào, các lớp ẩn và lớp đầu ra, dùng để học mối quan hệ giữa dữ liệu đầu vào và kết quả đầu ra.
```

---

## CNN là gì?

CNN là **Convolutional Neural Network**.

CNN dùng nhiều trong xử lý ảnh.

Vì sao?

```text
Ảnh có đặc trưng không gian: cạnh, góc, đường nét, hình dạng.
CNN có khả năng tự trích xuất các đặc trưng này.
```

Câu học thuộc:

```text
CNN là mạng nơ-ron tích chập, thường dùng trong xử lý ảnh vì có khả năng tự động trích xuất đặc trưng như cạnh, góc, đường nét và hình dạng đối tượng.
```

---

## Convolution là gì?

Convolution là **phép tích chập**.

Bạn hiểu đơn giản:

```text
Convolution dùng một bộ lọc nhỏ quét qua ảnh để tìm đặc trưng.
```

Ví dụ bộ lọc có thể phát hiện:

```text
Cạnh ngang
Cạnh dọc
Góc
Vùng sáng tối
```

Câu học thuộc:

```text
Convolution giúp trích xuất đặc trưng từ ảnh bằng cách dùng các bộ lọc quét qua ảnh.
```

---

## Pooling là gì?

Pooling dùng để **giảm kích thước ảnh** nhưng vẫn giữ thông tin quan trọng.

Ví dụ:

```text
Ảnh ban đầu lớn quá
Pooling làm ảnh nhỏ hơn
Mô hình chạy nhanh hơn
Giảm nhiễu và giảm overfitting
```

Câu học thuộc:

```text
Pooling giúp giảm kích thước dữ liệu, giữ lại đặc trưng quan trọng và làm mô hình tính toán nhanh hơn.
```

---

## RNN là gì?

RNN dùng cho dữ liệu chuỗi.

Ví dụ:

```text
Văn bản
Giọng nói
Dữ liệu thời gian
```

Câu học thuộc:

```text
RNN là mạng nơ-ron hồi quy, phù hợp với dữ liệu dạng chuỗi vì có khả năng ghi nhớ thông tin từ các bước trước đó.
```

---

## LSTM là gì?

LSTM là phiên bản cải tiến của RNN.

RNN thường khó nhớ thông tin dài. LSTM giúp khắc phục vấn đề đó.

Câu học thuộc:

```text
LSTM là cải tiến của RNN, giúp ghi nhớ thông tin dài hạn tốt hơn và xử lý tốt hơn các chuỗi dữ liệu dài.
```

---

## Transformer là gì?

Transformer dùng cơ chế **Attention**.

Nó là nền tảng của các mô hình như BERT, GPT.

Câu học thuộc:

```text
Transformer sử dụng cơ chế Attention để tập trung vào các phần quan trọng trong dữ liệu, thường được dùng nhiều trong xử lý ngôn ngữ tự nhiên.
```

---

## YOLO là gì?

YOLO dùng để **phát hiện đối tượng**.

Ví dụ:

```text
Phát hiện người
Phát hiện xe
Phát hiện biển báo
```

Câu học thuộc:

```text
YOLO là mô hình phát hiện đối tượng nhanh, có thể xác định vị trí và loại đối tượng trong ảnh.
```

---

## U-Net là gì?

U-Net dùng cho **phân đoạn ảnh**.

Phân đoạn ảnh nghĩa là chia ảnh theo từng vùng.

Ví dụ:

```text
Tách vùng khối u trong ảnh y tế
Tách đường, cây, xe trong ảnh
```

Câu học thuộc:

```text
U-Net là mô hình thường dùng cho phân đoạn ảnh, đặc biệt phổ biến trong ảnh y tế.
```

# 7. Dạng 1: Canny Edge Detection

## Đề có thể ra

```text
Cho ảnh rose.jpg.
Đọc ảnh và chuyển sang ảnh xám.
Áp dụng GaussianBlur với kernel (7, 7), sigma = 1.5.
Áp dụng Canny với ngưỡng 120 và 250.
Lưu ảnh kết quả.
```

## Code

```python
import cv2
import matplotlib.pyplot as plt

anh = cv2.imread("rose.jpg")

anh_xam = cv2.cvtColor(anh, cv2.COLOR_BGR2GRAY)

anh_mo = cv2.GaussianBlur(anh_xam, (7, 7), 1.5)

bien = cv2.Canny(anh_mo, 120, 250)

plt.imshow(bien, cmap="gray")
plt.title("Canny Edge")
plt.axis("off")
plt.show()

cv2.imwrite("edges_rose.jpg", bien)
```

## Bản chất

```text
GaussianBlur làm mịn ảnh, giảm nhiễu.
Canny phát hiện cạnh trong ảnh.
```

Câu nhận xét:

```text
GaussianBlur được dùng để làm mịn ảnh trước khi phát hiện cạnh, giúp giảm nhiễu. Canny Edge Detection giúp phát hiện các biên mạnh và yếu trong ảnh dựa trên hai ngưỡng cho trước.
```

# 8. Dạng 2: Sobel Edge Detection

## Đề có thể ra

```text
Cho ảnh mountain.jpg.
Đọc ảnh và chuyển sang ảnh xám.
Resize ảnh về 256 x 256.
Áp dụng Sobel theo trục X và Y.
Kết hợp hai kết quả.
Hiển thị và lưu ảnh.
```

## Code

```python
import cv2
import matplotlib.pyplot as plt

anh = cv2.imread("mountain.jpg")

anh_xam = cv2.cvtColor(anh, cv2.COLOR_BGR2GRAY)

anh_resize = cv2.resize(anh_xam, (256, 256))

sobel_x = cv2.Sobel(anh_resize, cv2.CV_64F, 1, 0)
sobel_y = cv2.Sobel(anh_resize, cv2.CV_64F, 0, 1)

sobel_x = cv2.convertScaleAbs(sobel_x)
sobel_y = cv2.convertScaleAbs(sobel_y)

ket_qua = cv2.addWeighted(sobel_x, 0.5, sobel_y, 0.5, 0)

plt.imshow(ket_qua, cmap="gray")
plt.title("Sobel Edge")
plt.axis("off")
plt.show()

cv2.imwrite("sobel_mountain.jpg", ket_qua)
```

## Bản chất

```text
Sobel X tìm biên theo phương ngang/dọc tùy chiều đạo hàm.
Sobel Y tìm biên theo phương còn lại.
Kết hợp Sobel X và Sobel Y để lấy biên tổng quát hơn.
```

Câu nhận xét:

```text
Sobel dùng để phát hiện biên theo hai hướng X và Y. Việc kết hợp Sobel X và Sobel Y giúp biểu diễn đầy đủ hơn các đường biên trong ảnh.
```

# 9. Dạng 3: Hough Transform phát hiện đường thẳng

## Đề có thể ra

```text
Cho ảnh street.jpg.
Đọc ảnh và chuyển sang ảnh xám.
Áp dụng Canny.
Phát hiện đường thẳng bằng HoughLinesP.
Vẽ đường thẳng lên ảnh gốc.
Lưu ảnh kết quả.
```

## Code

```python
import cv2
import matplotlib.pyplot as plt

anh = cv2.imread("street.jpg")

anh_xam = cv2.cvtColor(anh, cv2.COLOR_BGR2GRAY)

bien = cv2.Canny(anh_xam, 60, 180)

duong_thang = cv2.HoughLinesP(
    bien,
    1,
    3.14 / 180,
    threshold=120,
    minLineLength=50,
    maxLineGap=10
)

if duong_thang is not None:
    for line in duong_thang:
        x1, y1, x2, y2 = line[0]
        cv2.line(anh, (x1, y1), (x2, y2), (0, 255, 0), 2)

anh_rgb = cv2.cvtColor(anh, cv2.COLOR_BGR2RGB)

plt.imshow(anh_rgb)
plt.title("Hough Lines")
plt.axis("off")
plt.show()

cv2.imwrite("hough_street.jpg", anh)
```

## Bản chất

```text
Canny tìm biên trước.
Hough Transform dựa trên ảnh biên để tìm các đường thẳng.
cv2.line dùng để vẽ đường thẳng lên ảnh gốc.
```

Câu nhận xét:

```text
Hough Transform được sử dụng để phát hiện các đường thẳng trong ảnh. Trước khi áp dụng Hough, ảnh thường được phát hiện biên bằng Canny để làm nổi bật các đường biên.
```

# 11. Dạng 5: Bilateral Filter

## Đề có thể ra

```text
Cho ảnh road.jpg.
Đọc ảnh và chuyển sang ảnh xám.
Áp dụng Bilateral Filter.
Dùng Canny và Hough Transform để phát hiện đường thẳng.
Lưu ảnh kết quả.
```

## Code

```python
import cv2
import matplotlib.pyplot as plt

anh = cv2.imread("road.jpg")

anh_xam = cv2.cvtColor(anh, cv2.COLOR_BGR2GRAY)

anh_loc = cv2.bilateralFilter(anh_xam, 9, 75, 75)

bien = cv2.Canny(anh_loc, 90, 220)

duong_thang = cv2.HoughLinesP(
    bien,
    1,
    3.14 / 180,
    threshold=100,
    minLineLength=50,
    maxLineGap=10
)

if duong_thang is not None:
    for line in duong_thang:
        x1, y1, x2, y2 = line[0]
        cv2.line(anh, (x1, y1), (x2, y2), (0, 255, 0), 2)

anh_rgb = cv2.cvtColor(anh, cv2.COLOR_BGR2RGB)

plt.imshow(anh_rgb)
plt.title("Bilateral + Hough Lines")
plt.axis("off")
plt.show()

cv2.imwrite("hough_lines_road.jpg", anh)
```

## Bản chất

```text
Bilateral Filter làm mịn ảnh nhưng vẫn giữ biên tốt hơn Gaussian Blur.
```

Câu nhận xét:

```text
Bilateral Filter giúp làm mịn ảnh nhưng vẫn giữ lại các cạnh quan trọng, phù hợp khi cần giảm nhiễu trước khi phát hiện biên hoặc đường thẳng.
```
