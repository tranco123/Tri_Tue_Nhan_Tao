## Dạng 1: POS Tagging : POS là Part of Speech, nghĩa là từ loại.


### Đề : Nhập câu "The agile grey cat leaps above the sleeping mouse.". Sử dụng spaCy để xác định và hiển thị các danh từ, động từ và tính từ trong câu. Lưu kết quả vào file bai1.txt.

```python
import spacy
nlp = spacy.load("en_core_web_sm") 
cau = "The agile grey cat leaps above the sleeping mouse"
doc = nlp(cau)
for tu in doc : 
    if tu.pos_ in ["NOUN", "VERB", "ADJ", "PROPN"]:
        print(tu.text, tu.pos_)
# Lưu file 
with open("bai1.txt", "w", encoding="utf-8")as f:   # "w": chế độ ghi mới 
     for tu in doc : 
        if tu.pos_ in ["NOUN", "VERB", "ADJ", "PROPN"]:
            f.write(f"{tu.text}\t {tu.pos}\n") #\t: tạo khoảng cách 
```

## Dạng 2: Dependency Parsing:  Dependency Parsing là phân tích quan hệ phụ thuộc giữa các từ trong câu.
- Từ nào là trung tâm của câu?
- Từ nào bổ nghĩa cho từ nào?
- Chủ ngữ của động từ là từ nào?
- Tân ngữ là từ nào?


| Thành phần | Code           | Ý nghĩa                |
| ---------- | -------------- | ---------------------- |
| Child      | `tu.text`      | Từ đang xét            |
| Dep        | `tu.dep_`      | Quan hệ cú pháp        |
| Head       | `tu.head.text` | Từ mà nó phụ thuộc vào |

Thực hiện các yêu cầu sau:

    1 Sử dụng thư viện spaCy và mô hình en_core_web_sm để xử lý câu.
    2 Với mỗi từ trong câu, hiển thị:
        - Từ đang xét — text
        - Quan hệ cú pháp — dep_
        - Từ trung tâm mà nó phụ thuộc vào — head.text
    3 Vẽ cây phân tích cú pháp của câu bằng displacy.
    4 Lưu kết quả phân tích vào file bai2.txt.

```python
import spacy
nlp2 = spacy.load("en_core_web_sm")
cau2 = "The agile grey cat leaps above the sleeping mouse."
doc2 = nlp(cau)
for tu2 in doc2:
    print(tu2.text, tu2.dep_, tu2.head.text) 
```

```python
# ve cay theo cu phap 
from spacy import displacy
displacy.render(doc2, style="dep", jupyter=True) #style="dep": vẽ quan hệ phù hợp, jupyter=True → hiển thị trực tiếp trong notebook
```

```python
# lưu file: 
with open("bai2.txt", "w", encoding="utf -8") as f : 
    for tu2 in doc2:
        f.write(f"{tu2.text}\t{tu2.dep_}\t{tu2.head.text}\n")
```

## VI. Dạng 3: Named Entity Recognition – NER là nhận diện thực thể có tên.

| Thực thể           | Loại     |
| ------------------ | -------- |
| Queen Elizabeth II | Người    |
| United Kingdom     | Quốc gia |
| London             | Địa điểm |

| Nhãn     | Ý nghĩa                                     |
| -------- | ------------------------------------------- |
| `PERSON` | Người                                       |
| `ORG`    | Tổ chức                                     |
| `GPE`    | Quốc gia, thành phố hoặc địa danh chính trị |

#### Đề cho 1 đoạn văn : "Queen Elizabeth II was the longest-reigning monarch of the United Kingdom, coronated in London."
Thực hiện:
1. Sử dụng spaCy để xử lý đoạn văn.
2. Nhận diện các thực thể xuất hiện trong đoạn văn.
3. Hiển thị tên thực thể và loại thực thể.
4. Lưu kết quả vào file bai3.csv.

```python
import spacy
nlp = spacy.load("en_core_web_sm")
doan_van = "Queen Elizabeth II was the longest-reigning monarch of the United Kingdom, coronated in London."
doc = nlp(doan_van)
for thuc_the in doc.ents: # ent- Chứa danh sách thực thể mà spaCy nhận diện.
    print(thuc_the.text, thuc_the.label_) # label - lọa thực thể 
```

```python
from spacy import displacy
displacy.render(doc, style="ent", jupyter=True)
```

```python
# lƯU 
import pandas as pd
ket_qua = []
for thuc_the in doc.ents:
    ket_qua.append({"Thuc the": thuc_the.text,"Loai": thuc_the.label_,"Bat dau": thuc_the.start_char,"Ket thuc": thuc_the.end_char})
bang = pd.DataFrame(ket_qua)
bang.to_csv("bai3.csv", index=False)
```

## Dạng 4: Sentiment Analysis bằng VADER: Đây là bài toán phân tích xem câu mang cảm xúc

| Điểm       | Ý nghĩa               |
| ---------- | --------------------- |
| `pos`      | Mức tích cực          |
| `neg`      | Mức tiêu cực          |
| `neu`      | Mức trung tính        |
| `compound` | Điểm cảm xúc tổng hợp |

####  Đề bài:
Cho câu: I am very pleased with the product, but the price was too high.
Thực hiện các yêu cầu:
1. Sử dụng VADER để phân tích cảm xúc của câu.
2. Hiển thị các điểm:
    - Positive
    - Negative
    - Neutral
    - Compound
3. Dựa vào điểm Compound để kết luận câu mang cảm xúc tích cực, tiêu cực hay trung tính.
4. Lưu kết quả vào file bai4.csv.

```python
# cài thư viên phân tích cảm xúc nếu e cài xong rùi thì e ko cần chạy dòng này
%pip install nltk
import nltk
ket_qua = nltk.download(
    "vader_lexicon",
    download_dir=r"C:\Users\Admin\nltk_data",
    quiet=False,
    raise_on_error=True
)
print("Ket qua:", ket_qua)
```

```python
from nltk.sentiment.vader import SentimentIntensityAnalyzer
cau4 = "I am very pleased with the product, but the price was too high."
phan_tich = SentimentIntensityAnalyzer()
diem = phan_tich.polarity_scores(cau4)
print(diem)
```

```python
# kết luận cảm xúc 
compound = diem["compound"]
if compound >= 0.05:
    cam_xuc = "Tich cuc"
elif compound <= -0.05:
    cam_xuc = "Tieu cuc"
else:
    cam_xuc = "Trung tinh"
print("Cam xuc:", cam_xuc)
```

```python
# Lưu file 
import pandas as pd
bang = pd.DataFrame([{"Cau": cau,"Positive": diem["pos"],"Negative": diem["neg"],"Neutral": diem["neu"],"Compound": diem["compound"],"Cam xuc": cam_xuc}])
bang.to_csv("bai4.csv", index=False)
```
