# I. Bản chất của NLP là gì?

**NLP – Natural Language Processing** là xử lý ngôn ngữ tự nhiên.

Đầu vào thường là một câu hoặc đoạn văn:

```text
The agile grey cat leaps above the sleeping mouse.
```

Máy tính sẽ phân tích để biết:

* Từ nào là danh từ?
* Từ nào là động từ?
* Từ nào bổ nghĩa cho từ nào?
* Trong câu có tên người, tổ chức, địa điểm nào?
* Câu mang cảm xúc tích cực, tiêu cực hay trung tính?

Trong chương này, bạn chủ yếu dùng:

```text
spaCy → POS, Dependency Parsing, NER
VADER → Sentiment Analysis
```

spaCy cung cấp các thành phần xử lý như gán từ loại, phân tích phụ thuộc và nhận diện thực thể; mô hình tiếng Anh nhỏ `en_core_web_sm` bao gồm thông tin từ vựng, cú pháp và thực thể. ([spaCy][1])

# III. Khung chung của bài spaCy

Ba dạng POS, Dependency và NER đều mở đầu gần giống nhau:

```python
import spacy

nlp = spacy.load("en_core_web_sm")

cau = "The agile grey cat leaps above the sleeping mouse."

doc = nlp(cau)
```

## Giải thích

### `import spacy`

Gọi thư viện xử lý ngôn ngữ.

### `spacy.load("en_core_web_sm")`

Nạp mô hình tiếng Anh.

Trong đó:

```text
en = English
core = mô hình ngôn ngữ cốt lõi
web = huấn luyện trên văn bản web
sm = small, phiên bản nhỏ
```

### `doc = nlp(cau)`

Đưa câu vào mô hình.

`doc` chứa:

* Các từ trong câu.
* Từ loại.
* Quan hệ cú pháp.
* Thực thể.
* Vị trí ký tự.
